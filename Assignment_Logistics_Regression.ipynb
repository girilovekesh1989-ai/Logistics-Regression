```python
'''
Question 1 : What is Simple Linear Regression (SLR)? Explain its purpose.
'''
'''
Simple Linear Regression (SLR) is a statistical and machine learning method used to model the linear relationship between two continuous quantitative variables: a single independent (predictor) variable ($X$) and a dependent (response) variable ($Y$). [1, 2]
It fits a straight line ($y = \beta_0 + \beta_1X + \epsilon$) through the data points using the method of least squares, where $\beta_0$ is the $y$-intercept and $\beta_1$ is the slope of the line. This is detailed further in the [Scribbr Introduction to Simple Linear Regression](https://www.scribbr.com/statistics/simple-linear-regression/) and the [GeeksforGeeks Linear Regression Guide](https://www.geeksforgeeks.org/machine-learning/ml-linear-regression/). [3, 4]
## Purpose of Simple Linear Regression

* Predictive Modeling: Estimate or forecast future and unknown values of the dependent variable ($Y$) based on a given value of the independent variable ($X$).
* Relationship Analysis: Quantify and summarize the strength, direction, and nature of the linear association between two continuous variables.
* Explanatory Insights: Understand how much change in the response variable ($Y$) is expected on average for a one-unit change in the predictor variable ($X$). [1, 2, 5, 6, 7, 8]

'''
```

```python
'''
Question 2: What are the key assumptions of Simple Linear Regression?
'''
'''
The core assumptions of the simple linear regression model include:

* Linearity: The relationship between the independent variable ($X$) and the expected mean of the dependent variable ($Y$) is a straight line.
* Independence of observations: The error terms (residuals) are independent of each other with no autocorrelation (common in sequential or time-series data).
* Homoscedasticity (Constant variance): The variance of the residuals is the same and remains constant across all levels of the independent variable $X$.
* Normality of residuals: For any fixed value of $X$, the error terms or residuals are normally distributed (crucial for hypothesis testing, p-values, and confidence intervals).
* Zero conditional mean of errors: The average or expected value of the error term given any value of $X$ is equal to zero ($E(\epsilon \vert{} X) = 0$). [1, 2, 3]

'''
```

```python
'''
Question 3: Write the mathematical equation for a simple linear regression model and
explain each term.
'''
'''
The mathematical equation for a simple linear regression model is:
$$y = \beta_0 + \beta_1x + \varepsilon$$
Here is the explanation for each term in the model:

* $y$ (Dependent / Response Variable): This is the main variable you are trying to predict or explain.
* $x$ (Independent / Predictor Variable): This is the variable you use to predict or explain the changes in $y$.
* $\beta_0$ (Intercept): This represents the predicted value of $y$ when $x$ is equal to zero. On a graph, this is where the regression line crosses the vertical y-axis.
* $\beta_1$ (Slope Coefficient): This measures the change in $y$ for every one-unit increase in $x$. It tells you how steep the line is and the direction of the relationship (positive or negative).
* $\varepsilon$ (Error Term / Residual): This accounts for the randomness or variation in $y$ that cannot be explained by $x$. It captures the distance between the actual data points and the estimated regression line.

'''
```

```python
'''
Question 5: What is the method of least squares in linear regression?
'''
'''
The method of least squares in linear regression is a mathematical optimization technique used to find the best-fitting line through a set of data points by minimizing the sum of the squared vertical differences (residuals) between the observed values and the values predicted by the linear model. [1, 2]
------------------------------
## Core Concept

* Residuals (Errors): The difference between an actual observed value ($y_i$) and the predicted value ($\hat{y}_i$) on the line is $e_i = y_i - \hat{y}_i$.
* Squaring the Errors: Squaring each residual ensures that positive and negative errors do not cancel each other out, and it places a heavier penalty on larger errors.
* Minimizing the Sum: The goal is to make the total sum of squared errors ($S$) as small as possible: [1, 3, 4, 5, 6]
$$S = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

------------------------------
## Linear Model and Formulas
For a simple linear regression model of the form $\hat{y} = mx + c$ (or $y = a + bx$), the parameters—slope ($m$ or $b$) and y-intercept ($c$ or $a$)—are calculated using the [GeeksforGeeks Least Square Method Guide](https://www.geeksforgeeks.org/maths/least-square-method/) and [Investopedia Least Squares Guide](https://www.investopedia.com/terms/l/least-squares-method.asp) formulas: [1, 5, 7]

* Slope ($m$):
$$m = \frac{n \sum xy - (\sum x)(\sum y)}{n \sum x^2 - (\sum x)^2}$$
* Y-Intercept ($c$): [1]
$$c = \frac{\sum y - m \sum x}{n}$$

------------------------------
## Key Properties

* Unique Solution: For standard linear regression, taking the partial derivatives of the error function with respect to the slope and intercept and setting them to zero yields a direct, closed-form mathematical solution.
* Sensitivity to Outliers: Because errors are squared, large outliers have a disproportionately strong pull on the position of the regression line. [1, 3, 8]

'''
```

```python
'''
Question 6: What is Logistic Regression? How does it differ from Linear Regression?
'''
'''
Logistic Regression is a supervised machine learning algorithm used primarily for classification problems (such as determining whether an email is spam or not, or whether a patient has a disease). Instead of predicting an exact numeric value, it calculates the probability that a given input belongs to a specific category (typically a binary outcome like 0 or 1, Yes or No). [1, 2, 3, 4, 5]
It uses the sigmoid function (an S-shaped curve) to transform a linear combination of input features into a bounded probability value between 0 and 1. A threshold (commonly 0.5) is then applied to assign the probability to a discrete class. [1, 3]
------------------------------
## Differences Between Linear Regression and Logistic Regression

  Feature                               Linear Regression                                                Logistic Regression

  Primary Task                 Regression (predicting quantities)                       Classification (predicting categories/probabilities)
  Output Type          Continuous numerical values (e.g., house price, temperature)      Categorical / discrete values (e.g., 0/1, True/False)
  Target Distribution            Normal / Gaussian distribution                                                 Binomial distribution
  Underlying Function        Straight-line relationship ($y = mx + b$)                         S-shaped Sigmoid function ($P = \frac{1}{1 + e^{-z}}$)
  Estimation Method               Ordinary Least Squares (OLS)                                Maximum Likelihood Estimation (MLE)
  Decision Threshold                       Not required                                Required to map probabilities to discrete classes (e.g., $\ge 0.5$)
  Evaluation Metrics            $R^2$ Score, Mean Squared Error (MSE)                           Accuracy, Precision, Recall, AUC-ROC

'''
```

```python
'''
Question 7: Name and briefly describe three common evaluation metrics for regression
models.
'''
'''
Three common evaluation metrics for regression models are:

* Mean Absolute Error (MAE): Measures the average magnitude of the errors in a set of predictions, without considering their direction. It calculates the average of the absolute differences between actual and predicted values in the original units of the target variable. [1, 2, 3]
* Mean Squared Error (MSE): Measures the average of the squared differences between actual and predicted values. Because errors are squared, it heavily penalizes larger mistakes, making it more sensitive to outliers than MAE. [1, 2, 3]
* R-squared ($R^2$ Score): Also known as the coefficient of determination, it represents the proportion of variance in the dependent variable that is predictable from the independent variables. It typically ranges from 0 to 1, where higher values indicate a better fit of the model to the data. [1, 4, 5]

'''
```

```python
'''
Question 8: What is the purpose of the R-squared metric in regression analysis?
'''
'''
The purpose of the R-squared ($R^2$) metric, also known as the coefficient of determination, is to measure how well a regression model explains and predicts an outcome.
Specifically, it quantifies the proportion of the variance in the dependent variable (the target) that can be explained by the independent variables (the predictors) in the model.
## Key Details & Interpretation

* The Scale: $R^2$ values typically range from 0 to 1 (or 0% to 100%).
* 0 or 0%: The model explains none of the variability of the response data around its mean.
* 1 or 100%: The model explains all the variability, representing a perfect fit where all predicted values exactly match the observed data.

## Direct Comparison: $R^2$ vs. Adjusted $R^2$
While $R^2$ is highly useful, it has a limitation: it will always increase or stay the same when you add more variables, even if those variables add no real value. Because of this, data scientists often look at Adjusted R-squared as well.

       Metric                                Primary Purpose                                                             Behavior with Added Variables

  R-squared ($R^2$)              Measures the overall goodness-of-fit.                               Always increases or stays the same as more predictors are added.
  Adjusted $R^2$         Measures goodness-of-fit while penalizing unnecessary complexity.    Increases only if the new variable improves the model more than expected by chance; decreases otherwise. |


'''
```

```python
'''
Question 9: Write Python code to fit a simple linear regression model using scikit-learn
and print the slope and intercept.
(Include your Python code and output in the code box below.)
'''
'''
To fit a simple linear regression model and print its slope and intercept using scikit-learn, use the following code:
'''
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data (X must be 2D, y must be 1D)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([7, 14, 15, 18, 19])

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

# Extract slope and intercept
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (coef_): {slope}")
print(f"Intercept (intercept_): {intercept}")
```

```python
'''
Question 10: How do you interpret the coefficients in a simple linear regression model?
'''
'''
In a simple linear regression model ($Y = \beta_0 + \beta_1 X + \epsilon$), the coefficients $\beta_0$ (intercept) and $\beta_1$ (slope) are interpreted based on their mathematical and practical roles: [1, 2, 3]
## 1. The Intercept ($\beta_0$)

* Definition: The expected (predicted) value of the dependent variable $Y$ when the independent variable $X$ is equal to $0$.
* Practical Note: It may not always have a meaningful real-world interpretation (e.g., if $X$ is human height or weight, an $X$ of zero is physically impossible). [1, 2, 4]

## 2. The Slope Coefficient ($\beta_1$)

* Definition: The average or expected change in the dependent variable $Y$ for every one-unit increase in the independent variable $X$. [1, 2, 5]
* Direction (Sign):
* Positive ($\beta_1 > 0$): As $X$ increases, $Y$ tends to increase (direct relationship).
   * Negative ($\beta_1 < 0$): As $X$ increases, $Y$ tends to decrease (inverse relationship). [1, 2, 6]
* Magnitude (Absolute Value): Reflects the strength or rate of change in $Y$ per unit change in $X$. For example, as detailed in an [Outlier Article on Regression Coefficients](https://articles.outlier.org/coefficient-regression-formula), if the model is $Y = 3 + 2X$, a coefficient of $2$ means that for every 1-unit increase in $X$, $Y$ increases by an average of 2 units. [2, 7, 8]

'''
```
